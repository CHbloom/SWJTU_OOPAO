```python
from mamba_minimal import ModelArgs, Mamba
import torch
```

```python
def test():
    args = ModelArgs(d_model=64, n_layer=2, vocab_size=1000)
    model = Mamba(args)
    print(f"参数量: {sum(p.numel() for p in model.parameters()) / 1e6:.2f} M")

    x = torch.randint(0, args.vocab_size, (24, 257))  # (batch=2, len=32)
    logits = model(x)
    loss = logits.mean()  # 假装 loss
    loss.backward()
    print("前向+反向通过，logits 形状:", logits.shape)


if __name__ == "__main__":
    test()
```

```python
def test_gpu():
    device = "cuda" if torch.cuda.is_available() else "cpu"
    args  = ModelArgs(d_model=512, n_layer=4, vocab_size=1)
    model = Mamba(args).to(device)
    x     = torch.randint(0, args.vocab_size, (64, 256), device=device)
    out   = model(x)
    out.mean().backward()
    print("GPU 跑通 ✅  logits:", out.shape)


if __name__ == "__main__":
    test_gpu()
```
